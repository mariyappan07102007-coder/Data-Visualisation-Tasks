```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn  as sns
```

```python
df=pd.read_csv("/content/healthcare_dataset.csv")
df
```

```python
df['Medical_Code'].isnull()
```

```python
print(df['Medical_Code'].value_counts())
```

```python
df['Standardized_Medical_Code'] = df['Medical_Code'].str.replace('ICD-10-', '')
print(df[['Medical_Code', 'Standardized_Medical_Code']].head())
```

```python
df['Admission_Date'] = pd.to_datetime(df['Admission_Date'], format='%d-%m-%Y')
df['Discharge_Date'] = pd.to_datetime(df['Discharge_Date'], format='%d-%m-%Y')
df['Hospital_Stay_Days'] = (df['Discharge_Date'] - df['Admission_Date']).dt.days
display(df[['Admission_Date', 'Discharge_Date', 'Hospital_Stay_Days']].head())
```

```python
print("Summary Statistics for Billing Amount:")
display(df['Billing_Amount'].describe())
print("\nSummary Statistics for Hospital Stay Days:")
display(df['Hospital_Stay_Days'].describe())
```
